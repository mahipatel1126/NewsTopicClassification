```python
from transformers import DistilBertTokenizerFast, DistilBertForSequenceClassification

# Load tokenizer
tokenizer = DistilBertTokenizerFast.from_pretrained("best_model")

# Load model
model = DistilBertForSequenceClassification.from_pretrained("best_model")

print("✅ Model and tokenizer loaded successfully!")
```

```python
#Test a Real Prediction
import torch

# Labels used in your project
label_map = {
    0: "World",
    1: "Sports",
    2: "Business",
    3: "Sci/Tech"
}

# Sample news
text = "Apple launches a new AI chip for future MacBooks."

# Tokenize
inputs = tokenizer(
    text,
    return_tensors="pt",
    truncation=True,
    padding=True,
    max_length=128
)

# Prediction
with torch.no_grad():
    outputs = model(**inputs)

pred = torch.argmax(outputs.logits, dim=1).item()

print("News:")
print(text)
print("\nPrediction:", label_map[pred])
```
